## Numerical Differentiation
Numerical differentiation involves approximating the derivative of a function using finite differences. The most common methods are forward difference, backward difference, and central difference.

### Forward Difference
The forward difference approximation of the derivative of a function $f$ at a point $x$ is given by:
$$
f'(x) \approx \frac{f(x+h) - f(x)}{h}
$$

where $h$ is a small step size.

### Example: Approximating the Derivative
Let's approximate the derivative of $f(x) = x^2$ at $x = 2$ using the forward difference method.


```python
import numpy as np

# Define the function
def f(x):
    return x**2

# Define the forward difference approximation
def forward_difference(f, x, h=1e-5):
    return (f(x + h) - f(x)) / h

# Approximate the derivative at x = 2
x = 4
approx_derivative = forward_difference(f, x)
print(f'Approximate Derivative of f(x) at x = {x}: {approx_derivative:.4f}')
```

## Central Difference
The central difference approximation of the derivative of a function $f$ at a point $x$ is given by:
$$
f'(x) \approx \frac{f(x+h) - f(x-h)}{2h}
$$

### Example: Approximating the Derivative
Let's approximate the derivative of $f(x) = x^2$ at $x = 2$ using the central difference method.



```python
# Define the central difference approximation
def central_difference(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

# Approximate the derivative at x = 2
approx_derivative_central = central_difference(f, x)
print(f'Approximate Derivative (Central) of f(x) at x = {x}: {approx_derivative_central:.4f}')
```

## Numerical Partial Differentiation
Partial differentiation involves taking the derivative of a function with respect to one variable while holding the others constant. Numerical partial differentiation can be done using the same finite difference methods.

### Example: Approximating the Partial Derivative
Let's approximate the partial derivative of $f(x, y) = x^2 + y^2$ with respect to $x$ at $(x, y) = (2, 3)$ using the central difference method.


```python
# Define the function
def f_xy(x, y):
    return x**2 + y**2

# Define the partial derivative with respect to x
def partial_derivative_x(f, x, y, h=1e-5):
    return (f(x + h, y) - f(x - h, y)) / (2 * h)

# Approximate the partial derivative at (x, y) = (2, 3)
x, y = 2, 3
approx_partial_derivative_x = partial_derivative_x(f_xy, x, y)
print(f'Approximate Partial Derivative with respect to x at (x, y) = ({x}, {y}): {approx_partial_derivative_x:.4f}')
```

## Applications in Optimization
Differentiation is crucial in optimization, where we aim to find the maximum or minimum of a function. 

Numerical differentiation allows us to approximate gradients, which are used in gradient-based optimization algorithms like gradient descent.

Remember $max$ returns the value and $argmax$ returns the argument that produces the max value.

### Example: Gradient Descent
Let's implement a simple gradient descent algorithm to minimize the function $f(x) = x^2$.


```python
# Define the gradient of the function
def gradient(f, x, h=1e-5):
    return central_difference(f, x, h)

# Gradient descent algorithm
def gradient_descent(f, x_init, learning_rate=0.1, n_iterations=100):
    x = x_init
    for _ in range(n_iterations):
        grad = gradient(f, x)
        x -= learning_rate * grad
    return x

# Minimize f(x) starting from x = 10
x_init = 10
min_x = gradient_descent(f, x_init)
print(f'Minimum of f(x) found at x = {min_x:.4f}')
```
