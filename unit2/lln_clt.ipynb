# Law of Large Numbers (LLN) and Central Limit Theorem (CLT)
The Law of Large Numbers (LLN) and Central Limit Theorem (CLT) are two fundamental concepts in probability and statistics. These theorems explain how the distribution of sample averages approaches a normal distribution as the sample size becomes larger, and they form the basis for many statistical methods and inferences. In this lesson, we will explore LLN and CLT, understand their significance, and see examples of their applications.

## Law of Large Numbers (LLN)
The Law of Large Numbers states that as the size of a sample increases, the sample mean will get closer to the population mean. There are two forms of LLN: the Weak Law and the Strong Law.

### Definition
The Law of Large Numbers is expressed mathematically as:

$$
\lim_{n \to \infty} \frac{1}{n} \sum_{i=1}^{n} X_i = \mu
$$

where $X_i$ are independent and identically distributed random variables with expected value $\mu$.

### Assumptions
- The sample observations are independent.
- The sample observations come from the same distribution with a finite mean.

### Example: Coin Flips
Let's demonstrate LLN using coin flips. We'll see how the sample mean of the number of heads approaches the theoretical mean as the number of flips increases.


```python
import numpy as np
import matplotlib.pyplot as plt

# Number of coin flips
n_flips = 30
# Simulate coin flips (0: tails, 1: heads)
flips = np.random.binomial(1, 0.5, n_flips)

# Calculate cumulative mean
cumulative_mean = np.cumsum(flips) / np.arange(1, n_flips + 1)

# Plot the cumulative mean
plt.figure(figsize=(10, 6))
plt.plot(cumulative_mean, label='Cumulative Mean of Heads')
plt.axhline(0.5, color='red', linestyle='--', label='True Mean (0.5)')
plt.title('Law of Large Numbers: Coin Flips')
plt.xlabel('Number of Flips')
plt.ylabel('Cumulative Mean')
plt.legend()
plt.show()
```

## Central Limit Theorem (CLT)
The Central Limit Theorem states that the distribution of the sample means approaches a normal distribution as the sample size becomes large, regardless of the shape of the population distribution.

### Definition
The Central Limit Theorem is expressed mathematically as:

$$
\frac{\overline{X} - \mu}{\sigma / \sqrt{n}} \xrightarrow{d} N(0, 1)
$$

where $\overline{X}$ is the sample mean, $\mu$ is the population mean, $\sigma$ is the population standard deviation, and $n$ is the sample size.

### Assumptions
- The sample observations are independent.
- The sample observations come from the same distribution with a finite mean and variance.

### Example: Uniform Distribution
Let's demonstrate CLT using samples drawn from a uniform distribution. We'll see how the distribution of sample means approaches a normal distribution as the sample size increases.


```python
from scipy.stats import norm

# Parameters
n_samples = 1000
sample_size = 30

# Generate samples from a uniform distribution
samples = np.random.uniform(0, 1, (n_samples, sample_size))

# Calculate sample means
sample_means = np.mean(samples, axis=1)

# Plot the distribution of sample means
plt.figure(figsize=(10, 6))
plt.hist(sample_means, bins=30, density=True, alpha=0.7, label='Sample Means')
x = np.linspace(0, 1, 1000)
plt.plot(x, norm.pdf(x, np.mean(sample_means), np.std(sample_means)), 'r', label='Normal Distribution')
plt.title('Central Limit Theorem: Uniform Distribution')
plt.xlabel('Sample Mean')
plt.ylabel('Density')
plt.legend()
plt.show()
```

### Demonstration of CLT with Different Sample Sizes
Let's observe the effect of different sample sizes on the distribution of sample means.


```python
# Different sample sizes
sample_sizes = [5, 30, 100]
n_samples = 1000

plt.figure(figsize=(15, 5))
for i, size in enumerate(sample_sizes):
    # Generate samples from a uniform distribution
    samples = np.random.uniform(0, 1, (n_samples, size))
    
    # Calculate sample means
    sample_means = np.mean(samples, axis=1)
    
    # Plot the distribution of sample means
    plt.subplot(1, 3, i + 1)
    plt.hist(sample_means, bins=30, density=True, alpha=0.7, label=f'Sample Size = {size}')
    plt.plot(x, norm.pdf(x, np.mean(sample_means), np.std(sample_means)), 'r', label='Normal Distribution')
    plt.title(f'Sample Size = {size}')
    plt.xlabel('Sample Mean')
    plt.ylabel('Density')
    plt.legend()

plt.tight_layout()
plt.show()
```
