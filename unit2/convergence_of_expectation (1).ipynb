# Generating Draws from a Random Variable and Computing Expectation
In statistics and probability theory, the expectation (or mean) of a random variable is a measure of the central tendency of its probability distribution. 

In this lesson, we will learn how to generate draws from various random variables and compute their expectations. 

We will also demonstrate that the expectation for these distributions converges as the number of draws increases. 

When this is the case, we can use simulation to find the approximate expectation within some threshold $\epsilon$.

## Generating Draws and Computing Expectation

### Bernoulli Distribution
The Bernoulli distribution represents a random variable that has two possible outcomes: 1 (success) with probability $p$ and 0 (failure) with probability $1 - p$.

### Example: Computing Expectation for Bernoulli Distribution
The expectation for a Bernoulli random variable $X$ with parameter $p$ is given by:
$$
\mathbb{E}[X] = p
$$

Let's generate draws from a Bernoulli distribution and compute the expectation.


```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
p = 0.5  # Probability of success
n_draws = 10000  # Number of draws

# Generate Bernoulli distributed data
data = np.random.binomial(1, p, n_draws)

# Compute cumulative mean
cumulative_mean = np.cumsum(data) / np.arange(1, n_draws + 1)

# Plot the cumulative mean
plt.figure(figsize=(10, 6))
plt.plot(cumulative_mean[5:], label='Cumulative Mean')
plt.axhline(p, color='red', linestyle='--', label=f'True Mean (p={p})')
plt.title('Expectation for Bernoulli Distribution')
plt.xlabel('Number of Draws')
plt.ylabel('Cumulative Mean')
plt.legend()
plt.show()

# Final expectation
print(f'Final Expectation after {n_draws} draws: {cumulative_mean[-1]:.4f}')
```

### Binomial Distribution
The Binomial distribution represents the number of successes in a fixed number of independent Bernoulli trials with the same probability of success $p$.

### Example: Computing Expectation for Binomial Distribution
The expectation for a Binomial random variable $X$ with parameters $n$ (number of trials) and $p$ (probability of success) is given by:
$$
\mathbb{E}[X] = n \cdot p
$$

Let's generate draws from a Binomial distribution and compute the expectation.


```python
# Parameters
n = 10  # Number of trials
p = 0.5  # Probability of success
n_draws = 10000  # Number of draws

# Generate Binomial distributed data
data = np.random.binomial(n, p, n_draws)

# Compute cumulative mean
cumulative_mean = np.cumsum(data) / np.arange(1, n_draws + 1)

# Plot the cumulative mean
plt.figure(figsize=(10, 6))
plt.plot(cumulative_mean, label='Cumulative Mean')
plt.axhline(n * p, color='red', linestyle='--', label=f'True Mean (n*p={n * p})')
plt.title('Expectation for Binomial Distribution')
plt.xlabel('Number of Draws')
plt.ylabel('Cumulative Mean')
plt.legend()
plt.show()

# Final expectation
print(f'Final Expectation after {n_draws} draws: {cumulative_mean[-1]:.4f}')
```

### Normal Distribution
The Normal distribution is a continuous probability distribution characterized by its bell-shaped curve. It is defined by its mean $\mu$ and standard deviation $\sigma$.

### Example: Computing Expectation for Normal Distribution
The expectation for a Normal random variable $X$ with mean $\mu$ and standard deviation $\sigma$ is given by:
$$
\mathbb{E}[X] = \mu
$$

Let's generate draws from a Normal distribution and compute the expectation.


```python
from scipy.stats import norm

# Parameters
mu = 0  # Mean
sigma = 1  # Standard deviation
n_draws = 10000  # Number of draws

# Generate Normal distributed data
data = np.random.normal(mu, sigma, n_draws)

# Compute cumulative mean
cumulative_mean = np.cumsum(data) / np.arange(1, n_draws + 1)

# Plot the cumulative mean
plt.figure(figsize=(10, 6))
plt.plot(cumulative_mean, label='Cumulative Mean')
plt.axhline(mu, color='red', linestyle='--', label=f'True Mean (μ={mu})')
plt.title('Expectation for Normal Distribution')
plt.xlabel('Number of Draws')
plt.ylabel('Cumulative Mean')
plt.legend()
plt.show()

# Final expectation
print(f'Final Expectation after {n_draws} draws: {cumulative_mean[-1]:.4f}')
```

## Using Simulation to Approximate Expectation
When the expectation for a distribution converges, we can use simulation to find the approximate expectation within some threshold $\epsilon$.

### Example: Approximating Expectation for Exponential Distribution
The expectation for an Exponential random variable $X$ with rate parameter $\lambda$ is given by:
$$
\mathbb{E}[X] = \frac{1}{\lambda}
$$

Let's generate draws from an Exponential distribution and compute the expectation within a threshold $\epsilon$.


```python
from scipy.stats import expon

# Parameters
lam = 1  # Rate parameter (lambda)
n_draws = 10000  # Number of draws
epsilon = 0.01  # Threshold

# Generate Exponential distributed data
data = np.random.exponential(1/lam, n_draws)

# Compute cumulative mean
cumulative_mean = np.cumsum(data) / np.arange(1, n_draws + 1)

# Plot the cumulative mean
plt.figure(figsize=(10, 6))
plt.plot(cumulative_mean, label='Cumulative Mean')
plt.axhline(1/lam, color='red', linestyle='--', label=f'True Mean (1/λ={1/lam})')
plt.axhline(1/lam + epsilon, color='green', linestyle='--', label=f'Threshold (1/λ + ε)')
plt.axhline(1/lam - epsilon, color='green', linestyle='--')
plt.title('Expectation for Exponential Distribution')
plt.xlabel('Number of Draws')
plt.ylabel('Cumulative Mean')
plt.legend()
plt.show()

# Final expectation
print(f'Final Expectation after {n_draws} draws: {cumulative_mean[-1]:.4f}')

# Check if within threshold
within_threshold = np.abs(cumulative_mean[-1] - 1/lam) < epsilon
print(f'Expectation within threshold ε={epsilon}: {within_threshold}')
```
