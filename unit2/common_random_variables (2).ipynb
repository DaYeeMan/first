## Bernoulli Distribution
The Bernoulli distribution is a discrete probability distribution for a random variable which takes value 1 with success probability \( p \) and value 0 with failure probability \( 1 - p \).

### Assumptions
- There are only two possible outcomes: success (1) and failure (0).
- The probability of success \( p \) is the same for each trial.

### Application Example
- Modeling a coin flip, where success represents getting heads and failure represents getting tails.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli

# Parameters
p = 0.5  # Probability of success

# Generate Bernoulli distributed data
data = bernoulli.rvs(p, size=1000)

# Plot the distribution
plt.hist(data, bins=2, edgecolor='k', alpha=0.7)
plt.title('Bernoulli Distribution (p=0.5)')
plt.xlabel('Outcome')
plt.ylabel('Frequency')
plt.xticks([0, 1])
plt.show()
```

## Binomial Distribution
The Binomial distribution represents the number of successes in a fixed number of independent Bernoulli trials with the same probability of success \( p \).

### Assumptions
- There are a fixed number of trials \( n \).
- Each trial is independent.
- The probability of success \( p \) is the same for each trial.

### Application Example
- Modeling the number of heads in 10 flips of a fair coin.


```python
from scipy.stats import binom

# Parameters
n = 10  # Number of trials
p = 0.5  # Probability of success

# Generate Binomial distributed data
data = binom.rvs(n, p, size=1000)

# Plot the distribution
plt.hist(data, bins=n, edgecolor='k', alpha=0.7)
plt.title('Binomial Distribution (n=10, p=0.5)')
plt.xlabel('Number of successes')
plt.ylabel('Frequency')
plt.show()
```

## Geometric Distribution
The Geometric distribution represents the number of trials needed to get the first success in a series of independent Bernoulli trials with the same probability of success \( p \).

### Assumptions
- Each trial is independent.
- The probability of success \( p \) is the same for each trial.

### Application Example
- Modeling the number of coin flips needed to get the first heads.


```python
from scipy.stats import geom

# Parameters
p = 0.5  # Probability of success

# Generate Geometric distributed data
data = geom.rvs(p, size=1000)

# Plot the distribution
plt.hist(data, bins=max(data), edgecolor='k', alpha=0.7)
plt.title('Geometric Distribution (p=0.5)')
plt.xlabel('Number of trials to first success')
plt.ylabel('Frequency')
plt.show()
```

## Poisson Distribution
The Poisson distribution represents the number of events occurring in a fixed interval of time or space, given the average number of times the event occurs over that interval.

### Assumptions
- Events occur independently.
- The average rate (\( \lambda \)) at which events occur is constant.

### Application Example
- Modeling the number of emails received in an hour.

```python
from scipy.stats import poisson

# Parameters
lam = 3  # Average rate (lambda)

# Generate Poisson distributed data
data = poisson.rvs(lam, size=1000)

# Plot the distribution
plt.hist(data, bins=max(data)-min(data)+1, edgecolor='k', alpha=0.7)
plt.title('Poisson Distribution (λ=3)')
plt.xlabel('Number of events')
plt.ylabel('Frequency')
plt.show()
```

## Normal Distribution
The Normal distribution, also known as the Gaussian distribution, is a continuous probability distribution characterized by its bell-shaped curve. It is defined by its mean \( \mu \) and standard deviation \( \sigma \).

### Assumptions
- The data follows a bell-shaped symmetric distribution.
- The mean \( \mu \) is the center of the distribution.
- The standard deviation \( \sigma \) measures the spread of the distribution.

### Application Example
- Modeling the distribution of heights of adults in a population.


```python
from scipy.stats import norm

# Parameters
mu = 0  # Mean
sigma = 1  # Standard deviation

# Generate Normal distributed data
data = norm.rvs(mu, sigma, size=1000)

# Plot the distribution
plt.hist(data, bins=30, density=True, edgecolor='k', alpha=0.7)
x = np.linspace(mu - 3*sigma, mu + 3*sigma, 1000)
plt.plot(x, norm.pdf(x, mu, sigma), 'r')
plt.title('Normal Distribution (μ=0, σ=1)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.show()
```

## Uniform Distribution
The Uniform distribution is a continuous probability distribution where all intervals of the same length are equally probable.

### Assumptions
- The data is uniformly spread over a range \([a, b]\).

### Application Example
- Modeling the distribution of random numbers generated between 0 and 1.

Particularly useful for simulation and generating outcomes.

```python
from scipy.stats import uniform

# Parameters
a = 0  # Lower bound
b = 1  # Upper bound

# Generate Uniform distributed data
data = uniform.rvs(a, b-a, size=1000)

# Plot the distribution
plt.hist(data, bins=30, density=True, edgecolor='k', alpha=0.7)
x = np.linspace(a, b, 1000)
plt.plot(x, uniform.pdf(x, a, b-a), 'r')
plt.title('Uniform Distribution (a=0, b=1)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.show()
```

## Exponential Distribution
The Exponential distribution is a continuous probability distribution that describes the time between events in a Poisson process.

### Assumptions
- Events occur independently and at a constant average rate (\( \lambda \)).

### Application Example
- Modeling the time between arrivals of customers at a store.


```python
from scipy.stats import expon

# Parameters
lam = 1  # Rate (lambda)

# Generate Exponential distributed data
data = expon.rvs(scale=1/lam, size=1000)

# Plot the distribution
plt.hist(data, bins=30, density=True, edgecolor='k', alpha=0.7)
x = np.linspace(0, max(data), 1000)
plt.plot(x, expon.pdf(x, scale=1/lam), 'r')
plt.title('Exponential Distribution (λ=1)')
plt.xlabel('Time')
plt.ylabel('Density')
plt.show()
```
