## Loading and Displaying Images

```python
# Import necessary libraries
from PIL import Image
import matplotlib.pyplot as plt
import os

# Load an image
image_path = '/Users/romanpaolucci/Desktop/Saddle River Day/Courses (current)/Artificial Intelligence/Course Materials/Unit 3 - Computer Vision/3.1 Managing Images/fire.png'
image = Image.open(image_path)

# Display the image
plt.imshow(image)
plt.title('Original Image')
plt.axis('off')
plt.show()
```

### Cropping an Image

```python
# Define the crop box (left, upper, right, lower)
crop_box = (100, 100, 400, 400)
cropped_image = image.crop(crop_box)

# Display the cropped image
plt.imshow(cropped_image)
plt.title('Cropped Image')
plt.axis('off')
plt.show()
```

### Scaling an Image

```python
# Define the new size
new_size = (200, 200)
scaled_image = image.resize(new_size)

# Display the scaled image
plt.imshow(scaled_image)
plt.title('Scaled Image')
plt.axis('off')
plt.show()
```

### Saving Images

```python
# Save the cropped image
cropped_image_path = 'cropped.png'
cropped_image.save(cropped_image_path)

# Save the scaled image
scaled_image_path = 'scaled.png'
scaled_image.save(scaled_image_path)
```

### Batch Reading

```python
# Define the directory containing images
image_directory = '/Users/romanpaolucci/Desktop/Saddle River Day/Courses (current)/Artificial Intelligence/Course Materials/Unit 3 - Computer Vision/3.1 Managing Images'

# Initialize lists to store images and labels
images = []
labels = []

# Define a function to load images from a directory
def load_images_from_directory(directory):
    for filename in os.listdir(directory):
        if filename.endswith('.jpg') or filename.endswith('.png'):
            img_path = os.path.join(directory, filename)
            img = Image.open(img_path)
            images.append(img)
            
            # Either 1:
            # Assume labels are part of the filename, e.g., 'cat_01.jpg'
            #label = filename.split('_')[0]
            #labels.append(label)
            
            # Or 2:
            # Assume the directory has the same image labels
            labels.append('Fire')

# Load images and labels
load_images_from_directory(image_directory)

# Display the first image and its label
plt.imshow(images[0])
plt.title(f'Label: {labels[0]}')
plt.axis('off')
plt.show()
```
