```python
import numpy as np
```

```python
np.random.seed(42)

n_samples = 1000
n_features = 3

covariance = np.array([
    [1.0, 0.7, 0.5],
    [0.7, 1.0, 0.3],
    [0.5, 0.3, 1.0]
])

mean = np.zeros(n_features)
X = np.random.multivariate_normal(mean, covariance, n_samples)

X = X * 10 + 50
```

```python
np.cov(X, rowvar=False)
```

```python
X_Tilde = X - np.mean(X, axis=0)
m = X_Tilde.T @ X_Tilde / (999)
m
```

```python
eigenvalues, eigenvectors = np.linalg.eig(m)
eigenvalues, eigenvectors

eigenvalues = eigenvalues / np.sum(eigenvalues)
eigenvalues

pca_values = X_Tilde @ eigenvectors
pca_values
```
